#  Notes
- For XOR, the first layer expands the 2 inputs into more features (like 4) so the model has "room" to learn the complex logic.The final layer must end in out_features=1 to match the single answer ($0$ or $1$) you are trying to predict.

- The number of hidden features (like 4) determines the "learning capacity" of the model.
Too few neurons (1 or 2) make the problem mathematically impossible or very hard to solve; too many is a waste of computer resources. 4 is the "Sweet Spot" for XOR.

- MSE (Mean Squared Error) is like a "distance" ruler. It’s great for predicting continuous numbers like house prices ($400,000 vs $405,000).

- BCELoss is like a "truth" detector. It is mathematically designed for Yes/No (0 or 1) questions.

- The "Penalty" Difference: If the answer is 1 and the model guesses 0.01, MSE says "You're off by about 1." BCELoss says "You are extremely wrong!" and gives a massive penalty. This aggressive penalty helps the model learn the "sharp" logic of XOR much faster than MSE would.

- $0.1$ is a "fast" learning rate for simple problems. As the data gets more complex, we usually make the $lr$ smaller to be more precise. If $lr$ is 0.1: The model takes "giant leaps." It will find the answer in 500–1,000 epochs. If $lr$ is 10.0: The steps are too big. The model will "jump" over the correct answer and never find it (this is called "overshooting").

- In binary classification, a loss of 0.69 (specifically $-ln(0.5)$) means the model is randomly guessing. It has no idea what XOR is yet.
- Increasing the Learning Rate ($lr$) reduces the number of Epochs needed, provided the model stays stable.The final raw probabilities show the "confidence" of the model; 0.99 means the model is nearly certain, while 0.5 would mean it is completely guessing.

```python
import torch
import torch.nn as nn
```

```python
# The XOR dataset

X = torch.tensor([[0,0], [0,1], [1,0], [1,1]], dtype=torch.float32)
Y = torch.tensor([[0], [1], [1], [0]], dtype=torch.float32)
```

```python
epochs = 50
# 2-layer MLP: 2 inputs -> 4 hidden  -> 1 output

model = nn.Sequential(
    nn.Linear(2,4), # It projects those 2 numbers into a 4-dimensional space,  It tries to find 4 different "views" or patterns in your 2 input numbers.

    nn.ReLU(), # ReLU provides the "curve" needed to separate the 0s and 1s which are impossible to separate with a straight line.

    nn.Linear(4, 1), # It takes the 4 patterns found in the hidden layer and squashes them down into a single score.

    nn.Sigmoid() # An activation function that squashes any number into a range between 0 and 1. so If the result is 0.98, the model is saying "I'm 98% sure this is a 1.
)

loss_fn = nn.BCELoss() # Use MSE for predicting "How much?" (Regression). Use BCELoss for predicting "Which one?" (Classification).

optimizer = torch.optim.SGD(model.parameters(), lr=5)

# Training Loop
for epoch in range(epochs):
  pred = model(X)
  loss = loss_fn(pred, Y)

  optimizer.zero_grad()
  loss.backward()
  optimizer.step()

  if epoch % 10 == 0:
    print(f" Epoch {epoch} Loss: {loss.item():.4f}")


with torch.no_grad():
  raw_preds = model(X)
  print("\nRaw Probabilities (How sure the model is):")
  print(raw_preds)

  print("\nFinal Decided Classes:")
  print(raw_preds.round())
```
